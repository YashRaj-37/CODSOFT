```python
!pip install hmmlearn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import entropy
from hmmlearn import hmm
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/heart_failure_clinical_records_dataset.csv")
le = LabelEncoder()
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = le.fit_transform(df[col])

df.head()
```

```python
print(df.describe())
```

```python
print(df.corr()["DEATH_EVENT"].sort_values(ascending=False))

death_prob = df["DEATH_EVENT"].mean()
print(death_prob)
```

```python
joint_prob = len(df[(df["age"] > 60) & (df["DEATH_EVENT"] == 1)]) / len(df)
print(joint_prob)
```

```python
over60 = df[df["age"] > 60]
cond_prob = over60["DEATH_EVENT"].mean()
print(cond_prob)
```

```python
mle = df["DEATH_EVENT"].sum() / len(df)
print(mle)
p = 0.32
q = 0.20
kl = p*np.log(p/q) + (1-p)*np.log((1-p)/(1-q))
print(kl)
```

```python
features = df[["age","creatinine_phosphokinase","serum_creatinine"]].values
model = hmm.GaussianHMM(n_components=3, n_iter=100)
model.fit(features)
hidden_states = model.predict(features)
states = ["Healthy", "At Risk", "Critical"]
results = [states[i] for i in hidden_states]
print(results[:20])
print(model.transmat_)
```
